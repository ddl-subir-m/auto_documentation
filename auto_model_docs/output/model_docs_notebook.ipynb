## 3. Feature Engineering

The feature engineering strategy for this customer churn prediction model employs a progressive, multi-phase approach designed to systematically improve predictive performance while maintaining interpretability across different model architectures. The foundation begins with baseline raw features including customer demographics, service subscriptions, and contract details, then evolves through increasingly sophisticated transformations that capture complex behavioral patterns and risk indicators. This iterative methodology allows for controlled experimentation where each feature engineering version builds upon the previous one, enabling clear attribution of performance improvements to specific feature enhancements while supporting the diverse requirements of Logistic Regression, Random Forest, and XGBoost classifiers.

The first phase establishes a baseline using raw categorical and numerical features without transformation, providing a performance benchmark against which subsequent improvements are measured. The second phase introduces domain-informed derived features including tenure categorization to capture customer lifecycle stages, charge-per-tenure metrics to identify pricing efficiency, service count aggregations to measure customer engagement breadth, and high-value customer indicators based on monthly spending thresholds. These engineered features transform raw data into meaningful business signals that better represent customer behavior patterns and risk factors associated with churn.

The third and most advanced phase incorporates composite scoring mechanisms that synthesize multiple signals into unified risk indicators. The contract stability score evaluates commitment level by weighting contract types and tenure duration, while the payment risk score combines payment method characteristics with billing preferences to identify financially unstable customers. An engagement score aggregates adoption of value-added services such as online security, device protection, and streaming offerings, recognizing that customers with deeper service integration exhibit lower churn propensity. Additionally, this phase introduces categorical binning of monthly charges and binary service flags that enable both tree-based models to discover interaction effects and linear models to capture non-linear relationships through explicit feature representation.

The feature engineering pipeline implements careful handling of categorical variables through appropriate encoding strategies, with one-hot encoding applied to nominal features and ordinal encoding reserved for inherently ordered attributes like contract types. Numerical features are preserved in their original scale for tree-based models while standardization options are available for the Logistic Regression implementation. All engineered features are designed to be reproducible and deployable in production environments, with transformation logic encapsulated within the ChurnPipeline class to ensure consistency between training and inference. This comprehensive approach balances statistical rigor with business interpretability, creating features that not only improve model accuracy but also provide actionable insights for customer retention strategies.

## 4. Model Architecture

The customer churn prediction system employs a progressive model evaluation strategy that begins with interpretable baseline models and advances toward more sophisticated ensemble methods. Three distinct algorithms were selected to balance different objectives: Logistic Regression provides a transparent baseline with easily interpretable coefficients that help stakeholders understand the linear relationships between customer attributes and churn risk; Random Forest offers robustness to feature interactions and non-linear patterns while naturally handling the mixed categorical and numerical features present in telecom customer data; and XGBoost delivers state-of-the-art predictive performance through gradient boosting, particularly valuable for capturing complex decision boundaries in customer behavior patterns. This multi-model approach allows the team to validate that performance improvements from more complex models justify their reduced interpretability, a critical consideration when deploying machine learning systems that inform customer retention strategies.

The architectural design follows a deliberate three-phase experimentation framework tracked through MLflow's model registry, with each phase building upon insights from the previous iteration. The baseline phase establishes performance benchmarks using logistic regression with raw features, providing a simple yet effective model that serves as the interpretability standard. The feature engineering phase introduces Random Forest with enriched features including tenure categories, service engagement metrics, and high-value customer indicators, demonstrating how domain-specific transformations can improve model discrimination. The production phase deploys XGBoost with advanced composite features such as contract stability scores, payment risk indicators, and engagement metrics, achieving optimal performance through its ability to learn complex feature interactions and handle the inherent class imbalance in churn scenarios where only fifteen percent of customers actually leave.

Each model addresses the binary classification task with specific considerations for the business context and data characteristics. All three algorithms incorporate class balancing mechanisms—class_weight='balanced' for Logistic Regression and Random Forest, and scale_pos_weight for XGBoost—to ensure the models properly weight the minority churn class despite the 85-15 distribution in the training data. The models are evaluated across multiple metrics including precision, recall, F1-score, and ROC-AUC rather than accuracy alone, reflecting the business reality that false negatives (missing at-risk customers) and false positives (unnecessary retention offers) carry different operational costs. This comprehensive evaluation framework ensures that model selection considers both statistical performance and business impact, with particular attention to recall given the high cost of customer acquisition relative to retention.

The modular pipeline architecture enables seamless transitions between models while maintaining consistent feature engineering and prediction interfaces. The ChurnPipeline class abstracts the complexity of feature transformation and model inference, allowing business users to receive standardized outputs including binary predictions, probability scores, and risk level categorizations regardless of which underlying algorithm is deployed. ## 7. Validation Results

The validation strategy for this customer churn prediction system employs stratified k-fold cross-validation to ensure robust and unbiased performance estimates across all three model architectures: LogisticRegression, RandomForestClassifier, and XGBClassifier. Given the imbalanced nature of the dataset with a 15% churn rate, stratified sampling is critical to maintain consistent class distribution across all folds. This approach guarantees that each fold contains approximately the same proportion of churned and non-churned customers, preventing the model from being trained or validated on unrepresentative subsets of data that could lead to overly optimistic or pessimistic performance metrics.

The cross-validation process utilizes 5-fold stratified splitting, which divides the training dataset into five equal parts while preserving the target class ratio in each fold. During each iteration, four folds serve as the training set while the remaining fold acts as the validation set, and this process repeats five times so that every data point is used for validation exactly once. This methodology provides a more reliable estimate of model generalization performance compared to a single train-validation split, as it reduces the variance in performance metrics that can occur due to the particular random selection of validation samples. A fixed random seed of 42 is used throughout the validation process to ensure reproducibility of results across different experimental runs and model comparisons.

The choice of 5-fold cross-validation represents a practical balance between computational efficiency and statistical reliability for this dataset size of 10,000 samples. While 10-fold cross-validation could provide slightly more stable estimates, the computational overhead would be doubled, which becomes significant when training ensemble models like RandomForest and XGBoost with extensive hyperparameter tuning. With 5 folds, each validation set contains approximately 2,000 samples, providing sufficient data points to calculate reliable performance metrics while maintaining reasonable training times. This validation framework enables fair comparison across the three model architectures and the three feature engineering versions, as each model is evaluated under identical conditions.

The stratified cross-validation results are aggregated to produce mean performance metrics and standard deviations across all folds, offering insight into both the expected performance and the stability of each model. These metrics, including accuracy, precision, recall, F1 score, and ROC-AUC, are tracked in MLflow alongside the final holdout test set results. This dual evaluation approach ensures that model selection is based on consistent cross-validation performance while the final test set provides an unbiased estimate of real-world performance on completely unseen data. The stratification strategy is particularly important for this churn prediction task, as it ensures that minority class examples are adequately represented in both training and validation sets, enabling the models to learn effective decision boundaries for identifying at-risk customers.

The data preprocessing pipeline for this customer churn prediction system implements several critical transformations to ensure data quality and model readiness. Missing values in the total_charges field, which occur when customers have zero tenure months, are imputed with zero to reflect the absence of billing history. All categorical variables undergo label encoding to convert string values into numerical representations suitable for machine learning algorithms. The features senior_citizen, partner, dependents, phone_service, paperless_billing, and the target variable churned are already binary encoded as 0 and 1. Multi-class categorical features including multiple_lines, internet_service, online_security, online_backup, device_protection, tech_support, streaming_tv, streaming_movies, contract_type, and payment_method are systematically mapped to integer values, with special handling for "No internet service" and "No phone service" responses that are treated as distinct categories rather than missing data.

The feature engineering strategy evolved across three experimental phases to capture increasingly sophisticated customer behavior patterns. The baseline version (v1) utilized only raw features directly from the source data. Version 2 introduced derived features including tenure_category (binning customers into New, Mid-term, and Long-term segments based on months of service), charge_per_month_tenure (calculated as total_charges divided by tenure_months to identify spending patterns over time), service_count (aggregating the number of additional services subscribed), and high_value_customer (a binary flag for customers with monthly charges exceeding $70). The production version (v3) further enhanced the feature space with composite risk indicators: contract_stability_score combining contract type with tenure and payment method, payment_risk_score weighting payment method and billing preferences, engagement_score measuring service adoption breadth, monthly_charge_category for segmentation into Low, Medium, and High spending tiers, and individual service adoption flags. These engineered features were designed to capture non-linear relationships and interaction effects that simple raw features might miss.

Several important assumptions and limitations should be considered when deploying this model. The training data represents a snapshot in time with a 15% churn rate, and model performance may degrade if the underlying churn rate shifts significantly in production. The model assumes that historical patterns of customer behavior remain predictive of future churn, which may not hold during periods of major business changes such as pricing restructuring, competitive market shifts, or service quality disruptions. The categorical encoding approach assigns ordinal relationships to some inherently nominal variables, which could introduce spurious patterns, though tree-based models like Random Forest and XGBoost are relatively robust to this issue. The charge_per_month_tenure feature may be unstable for very new customers with low tenure values, potentially creating outliers that influence model predictions. Additionally, the model does not account for temporal trends or seasonality in churn behavior, as it treats all samples independently without considering time-series patterns.

Edge cases and data quality considerations include handling of customers with zero tenure months, who by definition have not yet had opportunity to churn but are included in the dataset with churned=0. Customers with "No internet service" responses for internet-dependent features are treated as a valid category rather than missing data, which appropriately captures their service profile but may conflate customers who actively declined internet service with those for whom it was unavailable. The class imbalance mitigation strategy using balanced class weights and scale_pos_weight parameters helps prevent the model from defaulting to predicting non-churn for all cases, but may result in higher false positive rates in production. Feature importance analysis from tree-based models should be interpreted cautiously, as correlated features like contract_type and tenure_months may have their importance split across multiple related variables. Finally, the model's predictions represent probabilities based on historical patterns and should be used as decision support tools rather than deterministic classifications, with threshold tuning recommended based on the specific business cost-benefit tradeoffs between retention intervention costs and customer lifetime value.
